## Generating some fake business datasets with lat, long coords

```python
import csv
import random

# Define the number of businesses you want in the dataset
num_businesses = 500

# Generate and write data to a CSV file
with open('berlin_businesses.csv', 'w', newline='') as csvfile:
    fieldnames = ['Business Name', 'Latitude', 'Longitude']
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()

    for _ in range(num_businesses):
        # Generate fake business name
        business_name = f'Business_{_}'

        # Generate random latitude and longitude coordinates for Berlin
        latitude = round(52.5200 + random.uniform(-0.1, 0.1), 6)
        longitude = round(13.4050 + random.uniform(-0.1, 0.1), 6)

        writer.writerow({'Business Name': business_name, 'Latitude': latitude, 'Longitude': longitude})

print(f'{num_businesses} fake businesses with latitude and longitude have been generated and saved to "berlin_businesses.csv".')

colors = ["Red", "Blue", "Green", "Yellow", "Purple", "Magenta", "Orange"]
names = [
    "Ramesh",
    "Baba",
    "Guntur",
    "Gilad",
    "Michel",
    "Vlad",
    "Horatio",
    "Leonor",
    "Mihailo",
    "Kat",
    "Maya",
    "Migel",
    "Hector",
    "Rabi",
    "Qurtaysh",
    "Maqbool",
    "Cyrus",
    "Alex",
    "Noor",
    "Alina",
    "Zainab",
    "Ksenia",
    "Sasha",
    "Leandra",
    "Latrine"
]

adjectives = [
    "Sweaty",
    "Nasty",
    "Stinky",
    "Crappy",
    "Shitty",
    "Abusive",
    "Cantankerous",
    "Addictive",
    "Slutty",
    "Unbelievable",
    "Rotten",
    "British",
    "Abhorrent",
    "Unforgivable",
    "Impressive",
    "Basic",
    "Delicious",
    "Toxic",
    "Disgusting",
    "Hilarious",
    "Ridiculous",
    "Pathetic",
    "Cowardly",
    "Humbug",
    "Intolerable"
]

locs = [
    "Deli",
    "Spaeti",
    "Whorehouse",
    "Shithouse",
    "Brewery",
    "Bar",
    "Kebabshop",
    "Kiosk",
    "Alehouse",
    "Meatshop"
]
```

```python
final_list = set()
while len(final_list) < 500:
    colorful = random.choice([True, False])
    name_gen = random.choice(names)
    adj_gen =random.choice(adjectives)
    loc_gen =random.choice(locs)
    if colorful:
        gen = f"{name_gen}'s {adj_gen} {random.choice(colors)} {loc_gen}"
    else:
        gen = f"{name_gen}'s {adj_gen} {loc_gen}"

    if gen not in final_list:
        final_list.add(gen)

final_list = list(final_list)

df = pd.read_csv("berlin_businesses.csv")
df["biz_name"] = final_list

temps = ["@gmail.com", "@yahoo.biz", "@outlook.com", "@gmx.de", "@mail.de"]
df["email_PRE"] = df.apply(lambda x: "".join([a[0] for a in x.biz_name.split()]).lower(), axis=1)
df["email"] = df.apply(lambda x: f"{x.email_PRE}{random.choice(temps)}", axis=1)

df[["biz_name", "Latitude", "Longitude", "email"]].to_csv("berlin_biz.csv", index=False)
```
